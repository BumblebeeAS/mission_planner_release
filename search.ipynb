```python
import numpy as np


def rotate_points(points, angle_rad, origin):
    """
    Rotate an array of 2D points by angle_rad around the given origin.
    """
    # Translate points to origin
    translated = points - origin
    # Rotation matrix
    c, s = np.cos(angle_rad), np.sin(angle_rad)
    R = np.array([[c, -s], [s, c]])
    # Apply rotation
    rotated = np.dot(translated, R.T)
    # Translate back
    return rotated + origin


def to_top_left(f, l, start_xy):
    # map to origin
    start_to_origin = np.array((-start_xy[0], -start_xy[1]))
    origin_to_top_left = np.array((f, -l))
    return start_to_origin + origin_to_top_left

def gen_square(f, b, l, r, start_xy):
    top_left = to_top_left(f, l, start_xy)
    btm_left = np.array((-f - b, 0))
    btm_right = np.array((0, l + r))
    top_right = np.array((f + b, 0))
    return np.array([top_left, btm_left, btm_right, top_right]).astype(float)
```

```python
points = gen_square(1, 1, 1, 1, (0, 0))
print("Points of the square:")
for idx, point in enumerate(points):
    print(f"Point {idx}: {point}")
end_point = np.sum(points, axis=0)
print(f"ended points 1 at {end_point}")

points = gen_square(2, 2, 2, 2, end_point)
for idx, point in enumerate(points):
    print(f"Point {idx + 4}: {point}")
print(f"ended points 2 at {np.sum(points,axis=0) + end_point}")
```

```python
import matplotlib.pyplot as plt


def accum_points(points, origin):
    accumulated = origin.copy()
    accumulated_points = [accumulated.copy()]
    for pt in points:
        accumulated += pt
        accumulated_points.append(accumulated.copy())
    return np.array(accumulated_points)

# Generate and plot first square and its vectors
origin = np.array((0, 0), dtype=float)

def gen_rotated_square(f, b, l, r, start_xy, angle_rad):
    points = gen_square(f, b, l, r, start_xy)
    return rotate_points(points, angle_rad, origin)

end = origin
for i in range(1, 4):
    offset = 0.2 * i
    # points = gen_rotated_square(offset, offset, offset, offset, origin, i * np.pi / 13)
    points = gen_square(offset, offset, offset, offset, end)
    rotated_accum = accum_points(points, end)
    end = np.sum(points, axis=0) + end
    plt.plot(rotated_accum[:, 1], rotated_accum[:, 0], color=f"C{i % 10}", marker="o", linestyle="-", label=f"Accumulated Path {i + 4}")


plt.axis("equal")
plt.grid(True)
plt.title("Squares and Vectors from Origin (XY Inverted)")
plt.xlabel("Y")
plt.ylabel("X")
plt.show()
```

```python
def _to_top_left(f: float, l: float, start_xy: list[float]) -> np.ndarray:
    """
    Convert forward and left distances to a 2D vector.
    Think of start_xy as some relative position/vector from the origin.
    We then apply the forward and left offsets to this position to get the position vector from the origin.
    """
    return np.array(
        [
            -start_xy[0] + f,
            -start_xy[1] - l,
        ],
        dtype=float,
    )


def _gen_square(f, b, l, r, start_xy):
    """
    Generate a square pattern with the given dimensions and starting position.
    The square is defined by its forward (f), backward (b), left (l),
    """
    top_left = _to_top_left(f, l, start_xy)
    btm_left = np.array((-f - b, 0))
    btm_right = np.array((0, l + r))
    top_right = np.array((f + b, 0))
    return np.array(
        [
            top_left,
            btm_left,
            btm_right,
            top_right,
        ],
        dtype=float,
    )
```

```python
end = np.zeros_like((2,1 ), dtype=float)
num_squares = 3
output_points = []
for i in range(1, num_squares + 1):
    offset = 0.2 * i
    points = _gen_square(offset, offset, offset, offset, end)
    output_points.append(points)
    end = np.sum(points, axis=0) + end

output_points = np.concatenate(output_points, axis=0)
print(output_points.shape)
print("Output points:")
for idx, point in enumerate(output_points):
    print(f"Point {idx}: {point}")
```
